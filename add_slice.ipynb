```python
import os
import numpy as np
from matplotlib import pyplot as plt
```

```python
train_dir = '../fdg/train_image/'
train_label_dir = '../fdg/train_label/'
```

```python
train_list=os.listdir(train_dir)
train_label_list=os.listdir(train_label_dir)

train_list.sort()
train_label_list.sort()
```

```python
print(train_list)
```

```python
for i in range(len(train_list)):
    print(i)
    #load image and label
    img=np.load(train_dir+train_list[i])
    label=np.load(train_label_dir+train_label_list[i])
    #shape
    height, width, depth, channels = img.shape
    if depth%16==0:
        h=0
    else:
        h=16-(depth%16)
    #pad_width
    pad_width = ((0,0), (0,0), (0, int(h)), (0, 0))
    pad_label_width = ((0,0), (0,0), (0, int(h)))
    
    #pad
    padded_array = np.pad(img, pad_width, mode='constant', constant_values=0)
    padded_label_array = np.pad(label, pad_label_width, mode='constant', constant_values=0)
    print(padded_array.shape)
    print(padded_label_array.shape)
    print(i)
    #save
    np.save((train_dir+train_list[i]), padded_array )
    np.save((train_label_dir+train_label_list[i]), padded_label_array )
```
