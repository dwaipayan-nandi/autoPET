```python
import torch
import numpy as np
import nibabel as nib
import os
import glob
import matplotlib.pyplot as plt
import random
```

```python
train_path="../fdg/image/"
label_path="../fdg/label/"
```

```python
train_list = os.listdir(train_path)
label_list = os.listdir(label_path)
```

```python
train_list
```

```python
save_train="../fdg/train_image/"
save_label="../fdg/train_label/"
```

```python
j=0
for i in range (0,2028,2):
    img_0 = nib.load(train_path+train_list[i]).get_fdata()
    img_1 = nib.load(train_path+train_list[i+1]).get_fdata()
    n0 = (img_0-img_0.min())/(img_0.max()-img_0.min())
    n1 = (img_1-img_1.min())/(img_1.max()-img_1.min())
    n = np.stack((n0, n1), axis=-1)
    n = n.astype('float16')
    np.save((save_train+"pat_"+str(j)+".npy"), n )
    os.remove(train_path+train_list[i])
    os.remove(train_path+train_list[i+1])
    print(j)
    j=j+1
```

```python
j=0
for i in range(0,1014):
    label=nib.load(label_path+label_list[i]).get_fdata()
    label = label.astype('float16')
    np.save((save_label+"pat_"+str(j)+".npy"), label)
    os.remove(label_path+label_list[i])
    print(j)
    j=j+1
```
