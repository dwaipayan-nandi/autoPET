```python
import numpy as np
import os
from patchify import patchify
from matplotlib import pyplot as plt
```

```python
train_img_dir = '../fdg/train_image/'
label_img_dir = '../fdg/train_label/'
```

```python
train_img_list = os.listdir(train_img_dir)
label_img_list = os.listdir(label_img_dir)
```

```python
for i in range(0, len(train_img_list)):
    img = np.load(train_img_dir+train_img_list[i])
    label = np.load(label_img_dir+label_img_list[i])
    #print(img.shape)
    channel_0=img[:,:,:,0]
    channel_1=img[:,:,:,1]
    patches_0 = patchify(channel_0, (400,400,16), step=16)
    patches_1 = patchify(channel_1, (400,400,16), step=16)
    patches_label = patchify(label, (400,400,16), step=16) 
    patch=np.stack((patches_0, patches_1), axis=-1)
    print(patch.shape)
    print(patches_label.shape)
    print(i)
    sum=0
    for l in range (0,(patch.shape)[2]):
        mini_img = patch[:,:,l,:,:,:,:]
        mini_label= patches_label[:,:,l,:,:,:]
        mini_img=np.squeeze(mini_img)
        mini_label=np.squeeze(mini_label)
        #print((train_img_list[i])[:-4])
        x = (train_img_dir+(train_img_list[i])[:-4]+'_'+str(l)+'.npy')
        y = (label_img_dir+(label_img_list[i])[:-4]+'_'+str(l)+'.npy')
        np.save((x), mini_img)
        np.save((y), mini_label)
        sum=sum+1
    print("SAVED")
    if sum==(patch.shape)[2]:
       os.remove(train_img_dir+train_img_list[i])
       os.remove(label_img_dir+label_img_list[i])
       print("DELETED")
    print("*********************")
```
